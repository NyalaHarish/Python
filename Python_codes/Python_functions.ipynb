```python
def factorial(n):
    if n==0 or n==1:
        return 1
    else:
        return n*factorial(n-1)

print(factorial(5))
```

```python
def blastoff(n):
    if n<=0:
        print("Blast off!")
    else:
        print(n)
        blastoff(n-1)

```

```python
print(blastoff(5))
```

```python
class book:
    def __init__(self,title,author):
        self.title=title
        self.author=author
        self.checked_out=False
    def check_out(self):
        if not self.checked_out:
            self.checked_out=True
            print(f"{self.title} is checked out")
        else:
            print(f"{self.title} is already checked out")

    def return_book(self):
        if self.checked_out:
            self.checked_out=False
            print(f"{self.title} is returned back")
        else:
            print(f"{self.title} was not checked out")
            
```

```python
class member:
    def __init__(self,name,mem_id):
        self.name=name
        self.mem_id=mem_id
        self.borrowed_books=[]
    def borrow_book(self,book):
        if len(self.borrowed_books)<5:
            if not book.check_out:
                book.check_out()
                self.borrowed_books.append(book)
                print(f"{self.name} has borrowed {book.title}")
            else:
                print(f"{book.title} is already checked out")
        else:
            print(f"{self.name} cannot borrow more than five books")
    def return_book(self,book):
        if book in self.borrowed_books:
            book.return_book()
            self.borrowed_books.remove(book)
            print(f"{book.title} was returned by {self.name}")
        else:
            print(f"{self.name} didn't borrowed {book.title}")
        
  

```
