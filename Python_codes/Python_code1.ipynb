```python
# The first 10 multiples of n, each printed on a new line.

x=int(input())
i=1

while i<=10:
    print(i*x)
    i+=1
```

```python
# Code to sum of all odd from 1 to 'n'
n=int(input())
sum=0
for i in range(1,n+1):
    if i%2==1:
        sum+=i

print(sum)
```

```python
#Prints the difference between sum of odd and sum of even numbers of givrn 10 numbers.
i=0
sum_eve=0
sum_odd=0
while i<10:
    x=int(input())
    if x%2==0:
        sum_eve+=x
    else:
        sum_odd+=x
    i+=1
diff=sum_odd-sum_eve

print(diff)
```

```python
#print the factorial of a given number.
n=int(input())
fact=1
for i in range(1,n+1):
    fact=i*fact
print(fact)
```

```python
#write a program that determines whether the triangle formed by the sides exists
X = int(input())
Y = int(input())
Z = int(input())
if X+Y<=Z or X+Z<=Y or Y+Z<=X:
    print("Not a Triangle")   
else: 
    if X!=Y and Y!=Z and X!=Z:
        print("Scalene Triangle")
    elif X==Y and Y==Z:
        print("Equilateral Triangle")
    elif X==Y or Y==Z or X==Z:
        print("Isosceles Triangle")
```

```python
# Check if the string is a palindrome
string = input()
lst1=list(string)
lst2=lst1[::-1]
if lst1==lst2:
    print("The string is a palindrome.")
else:
    print("The string is not a palindrome.")
```
