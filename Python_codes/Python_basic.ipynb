```python
def group_words_length(sentence):
    words=sentence.split()
    grouped_words={}
    for word in words:
        length=len(word)
        if length not in grouped_words:
            grouped_words[length]=[]
            print(length)
        grouped_words[length].append(word)

    return grouped_words


sentence="This is exanple for words with grouped length"
result=group_words_length(sentence)

print(result)
print("Grouped words by length:")
for length,group in result.items():
    print(f"Length {length}:{group}")

        
```
