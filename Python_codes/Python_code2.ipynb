```python
b=input()
up=False
lo=False
nu=False
sp=False

for char in b:
    if char.isdigit():
        nu=True
    elif char.isupper():
        up=True
    elif char.islower():
        lo=True
    elif not char.isalnum():
        sp=True

if up and lo and nu and sp:
    print("Valid")
else:
    print("Invalid")
```

```python
class Book:
    def __init__(self,title,author):
        self.title=title
        self.author=author
        self.checked_out=False
    def check_out(self):
        if not self.checked_out:
            self.checked_out=True
            print(f"{self.title} is checked out now")
        else:
            print(f"{self.title} is already checked out")
    def return_book(self):
        if self.checked_out:
            self.checked_out=False
            print(f"{self.title} is returned")
        else:
            print(f"{self.title} is not checked out")

```

```python
class Vehicle:
    def __init__(self,make,model,year):
        self.make=make
        self.model=model
        self.year=year
    def display(self):
        print(f"{self.make} made {self.model} in {self.year}")
    def start(self):
        print(f"{self.model} is starting")
```

```python
class car(Vehicle):
    def __init__(self,make,model,year,door_num):
        super().__init__(make,model,year)
        self.door_num=door_num
    def display_car(self):
        self.display()
        print(f"number of doors are {self.door_num}")
class bike(Vehicle):
        def __init__(self,make,model,year,bike_type):
            super().__init__(make,model,year)
            self.bike_type=bike_type
        def display_bike(self):
            self.display()
            print(f"bike model is {self.bike_type}")
```

```python
car1=car("Honda","Civic",2022,4)
bike1=bike("Bajaj","Pulsar",2021,"NS")
```

```python
car1.display()
```

```python
bike1.display()
```

```python
class shape:
    def area(self):
        pass
class circle(shape):
    def __init__(self,radius):
        self.radius=radius
    def area(self):
        return 3.14*self.radius*self.radius

class rectangle(shape):
    def __init__(self,width,breadth):
        self.width=width
        self.breadth=breadth
    def area(self):
        return self.width*self.breadth
        


        
```

```python
def print_area(shape):
    print(f"The area is {shape.area()}")
```

```python
cir=circle(5)
rect=rectangle(4,5)
```

```python
cir.area()
```

```python
rect.area()
```

```python
class person:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def get_name(self):
        return self.name
    def set_name(self,name):
        if isinstance(name,str) and name:
            self.name=name
        else:
            print("Invalid name")
    def get_age(self):
        return self.age
    def set_age(self,age):
        if isinstance(age,int) and age>0:
            self.age=age
        else:
            print("Invalid age")
```
